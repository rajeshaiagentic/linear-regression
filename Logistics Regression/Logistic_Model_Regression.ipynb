### Import nasseri

```python
import pandas as pd
import warnings
warnings.filterwarnings('ignore')
```

### Data Import

```python
Health_Data = pd.read_csv('framingham.csv')
Health_Data
```

### Data Understanding

```python
Health_Data.shape
```

```python
Health_Data.isna().sum()
```

```python
Health_Data.dtypes
```

### Data 582 data not properly so I will Droped

```python
Health_Data.dropna(inplace=True)
Health_Data
```

### Data Peparasion

```python
X = Health_Data.drop(labels='TenYearCHD',axis=1)
X
```

```python
y = Health_Data['TenYearCHD']
y
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42, shuffle=True, stratify=y)
```

```python
X_train.shape, y_train.shape
```

```python
X_test.shape, y_test.shape
```

### Model Bulding

```python
from sklearn. linear_model import LogisticRegression
```

```python
logistic_model = LogisticRegression()
```

### Model Traning

```python
logistic_model.fit(X_train,y_train)
```

```python
logistic_model.fit(X_test,y_test)
```

### Model Testing

```python
y_train_pred = logistic_model.predict(X_train)
y_train_pred
```

```python
y_test_pred = logistic_model.predict(X_test)
y_test_pred
```

### Model_Eveluvation Metrics

```python
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score, roc_auc_score
```

### 1.1 Traning Data

```python
print(classification_report(y_train,y_train_pred))
```

```python
confusion_matrix(y_train,y_train_pred)
```

```python
accuracy_score(y_train,y_train_pred)
```

```python
roc_auc_score(y_train,y_train_pred)
```

### 1.2 Testing Data

```python
print (classification_report(y_test,y_test_pred))
```

```python
confusion_matrix(y_test,y_test_pred)
```

```python
accuracy_score(y_test,y_test_pred)
```

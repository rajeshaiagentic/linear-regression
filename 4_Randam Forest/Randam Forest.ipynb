### Import Nessaries

```python
import pandas as pd
import warnings
warnings.filterwarnings('ignore')
```

### Data Collection 

```python
loan_data = pd.read_csv("loan_approval_dataset.csv")
loan_data
```

### Data Peparasion

```python
loan_data.shape
```

```python
loan_data.isna().sum()
```

```python
loan_data.dtypes
```

```python
loan_data.drop(labels='loan_id',axis=1,inplace=True)
```

```python
loan_data
```

```python
### education and self_employed, loan_status Fetures srt so convert number Ml only understand numbers 

loan_data[' education']     = loan_data[' education'].replace(to_replace=[' Graduate',' Not Graduate'], value=[1,0],)
loan_data[' self_employed'] = loan_data[' self_employed'].replace(to_replace=[' Yes',' No'],value=[1,0])
loan_data[' loan_status'] = loan_data[' loan_status'].replace(to_replace=[' Approved',' Rejected'],value=[1,0])
```

```python
loan_data
```

```python
loan_data.dtypes
```

```python
X = loan_data.drop(labels=[' loan_status'], axis=1)
X
```

```python
y = loan_data[' loan_status']
y
```

```python
from sklearn. model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42, shuffle=True)
```

```python
X_train.shape, y_train.shape
```

```python
X_test.shape,y_test.shape
```

### Model Bulding

```python
from sklearn. ensemble import RandomForestClassifier
from sklearn.model_selection import GridSearchCV
```

```python
random_froest = RandomForestClassifier(n_estimators=30,max_depth=9)
```

```python
g_searchcv = GridSearchCV(estimator=random_froest,param_grid={'n_estimators':[10,25,30,40,50,60,70,80,90,100],
                                                              'max_depth':[3,4,5,6,7,8,9,10]})
g_searchcv.fit(X_train,y_train)
```

```python
print(g_searchcv.best_params_)
print(g_searchcv.best_score_)
```

### Model Traning

```python
random_froest.fit(X_train,y_train)
```

### Model Testing

```python
y_pred_training = random_froest.predict(X_train)
y_pred_training
```

```python
y_pred_testing = random_froest.predict(X_test)
y_pred_testing
```

### Model Evelovation_Metric

```python
from sklearn. metrics import confusion_matrix, classification_report, accuracy_score, roc_auc_score
```

### Model Traning

```python
confusion_matrix(y_train,y_pred_training)
```

```python
print(classification_report(y_train,y_pred_training))
```

```python
accuracy_score(y_train,y_pred_training)
```

```python
roc_auc_score(y_train,y_pred_training)
```

### Model Testing

```python
confusion_matrix(y_test,y_pred_testing)
```

```python
print(classification_report(y_test,y_pred_testing))
```

```python
accuracy_score(y_test,y_pred_testing)
```

```python
roc_auc_score(y_test,y_pred_testing)
```

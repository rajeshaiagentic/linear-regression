### Data Assumption

```python
students_data.corr()
```

### data preparation

```python
X = students_data.drop(labels = 'Performance Index',axis = 1)
X
```

```python
y = students_data['Performance Index']
y
```

```python
from sklearn. model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42,shuffle=True)
```

```python
X_train.shape, y_train.shape
```

```python
X_test.shape, y_test.shape
```

### Model Building & Calling

```python
from sklearn.linear_model import LinearRegression
```

```python
Linear_model = LinearRegression()
```

### Model Traning

```python
Linear_model.fit(X_train,y_train)
```

```python
Linear_model.fit(X_test,y_test)
```

### Model Testing

```python
y_pred_train = Linear_model.predict(X_train)
print(y_pred_train)
```

```python
y_pred_testing = Linear_model.predict(X_test)
print(y_pred_testing)
```

### Model Evaluation Metrics

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, root_mean_squared_error, r2_score
```

```python
mean_absolute_error(y_train,y_pred_train),mean_squared_error(y_train,y_pred_train),root_mean_squared_error(y_train,y_pred_train),r2_score(y_train,y_pred_train)
```

```python
mean_absolute_error(y_test,y_pred_testing),mean_squared_error(y_test,y_pred_testing),root_mean_squared_error(y_test,y_pred_testing),r2_score(y_test,y_pred_testing)
```

### Very Important Metrics adjacent r2 Score Checking

### column names space then added Q 

```python
linear_stats_model = sms.ols(
    formula='Q("Performance Index") ~ Q("Hours Studied") + Q("Previous Scores") + Q("Extracurricular Activities") + Q("Sleep Hours") + Q("Sample Question Papers Practiced") ',
    data=students_data
).fit()
```

```python
linear_stats_model. rsquared_adj
```

```python
y_states_test = linear_stats_model.predict(X_test)
y_states_test
```

```python
r2_score(y_test,y_states_test)
```
